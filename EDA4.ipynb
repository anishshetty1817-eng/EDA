```python
import numpy as np

a = np.random.randint(1, 10, (2, 3))
b = np.random.randint(1, 10, (2, 3))

print(a)
print(b)

vertical1 = np.vstack((a, b))
vertical2 = np.concatenate((a, b), axis=0)

print(vertical1)
print(vertical2)

print(np.array_equal(vertical1, vertical2))

```

```python
horizontal1 = np.hstack((a, b))
horizontal2 = np.concatenate((a, b), axis=1)

print(horizontal1)
print(horizontal2)

print(np.array_equal(horizontal1, horizontal2))
```

```python
import numpy as np
x = np.arange(1, 16)

print(x)

parts = np.split(x, 3)

print(parts[0])
print(parts[1])
print(parts[2])
```

```python
import numpy as np

x = np.random.randint(1, 100, 15)
print(x)
p = np.partition(x, 4)

print(p[:4])
s = np.sort(x)

print(s[:4])
```

```python
import numpy as np

x = np.arange(1, 25).reshape(6, 4)
print(x)
horizontal = np.hsplit(x, 2)

print(horizontal[0])
print(horizontal[1])
```

```python
vertical = np.vsplit(x, 2)
print(vertical[0])
print(vertical[1])
```
