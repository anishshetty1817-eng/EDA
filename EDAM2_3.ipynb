```python
import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('Karnataka', 2010),
    ('Karnataka', 2020),
    ('Maharashtra', 2010),
    ('Maharashtra', 2020),
    ('Tamil Nadu', 2010),
    ('Tamil Nadu', 2020)
], names=['State', 'Year'])

population = pd.Series(
    [61100000, 67500000, 112000000, 124000000, 72100000, 77800000],
    index=index,
    name='Population'
)

print(population)
```

```python
print(population.loc['Karnataka'])

print(population.loc[('Tamil Nadu', 2020)])
```
