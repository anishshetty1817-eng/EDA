```python
#5

import numpy as np

student_names = np.array(["Anish", "Haneesh", "Basanth", "Vishwa"])
scores = np.array([88, 72, 95, 64])

index = np.argsort(scores)

ranked_students = student_names[index]

print(ranked_students)
```
